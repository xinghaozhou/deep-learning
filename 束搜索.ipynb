```python
# 贪心搜索
# 在seq2seq中我们使用了贪心搜索来预测序列
#   - 将当前时刻预测概率最大的词输出
# 但贪心可能不是最优的
# 因为每次新的输出会导致不同的下一个预测

# 束搜索
#   - 保存最好的k个候选
#   - 在每个时刻, 对每一个候选新加一项(n个可能), 在kn个选项中选出最好的k个
# 比如 k =2,
# 我们在每一层都选两个, 这里的层是并行的, 也就是说一直就有两个候选
# 时间复杂度O(knt) 因为在每一个时刻做kn次搜索
# 每个候选的最终分数是:
# 1/L^a log p(y1,...,yL) = 1/L^a Σ^L_t(log p(y_t'| y1,...,y_t'-1))
# 因为连乘中, 句子越长越小, 所以我们用log
# 1/L^a 是给长句子一点好处

# 总结
# 束搜索在每次搜索时保存k个最好的候选
# k = 1时是贪心搜索
# k = n时是束搜索, etc n= 5, 10
```
