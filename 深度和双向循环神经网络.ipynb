```python
# 深度循环神经网络
# 回顾循环神经网络
# 在更新隐藏状态: h_t = ϕ(W_hh * h_t-1 + W_hx * x_t-1 + b_h) 中如何得到更多非线性
# 回答: 更深, 多加几个隐藏层变成深RNN

# 第一层: (H_t)^1 = f_1((H_t-1)^1,  X_t)
# 第j层: (H_t)^j = f_j((H_t-1)^j, (H_t)^j-1) # 就是这个隐藏层的上一个时间和上一个隐藏层的这个世界
# 输出: O_t = g(H_t)^L 最后一层的隐藏状态
```

```python
# 总结
# 深度循环神经网络使用多个隐藏层来获得更多非线性性
```

```python
import torch
from torch import nn
from d2l import torch as d2l

batch_size, num_steps = 32, 35
train_iter, vocab = d2l.load_data_time_machine(batch_size, num_steps)

# 通过设置num_layers的值来设定隐藏层
vocab_size, num_hiddens, num_layers = len(vocab), 256, 2
num_inputs = vocab_size
device = d2l.try_gpu()
lstm_layer = nn.LSTM(num_inputs, num_hiddens, num_layers)
model = d2l.RNNModel(lstm_layer, len(vocab))
model = model.to(device)

num_epochs, lr = 500, 2
d2l.train_ch8(model, train_iter, vocab, lr, num_epochs, device)
```

```python
# 这里是一个错误案例, 我们不能用来预测
import torch
from torch import nn
from d2l import torch as d2l

batch_size, num_steps, device = 32, 35, d2l.try_gpu()
train_iter, vocab = d2l.load_data_time_machine(batch_size, num_steps)
vocab_size, num_hiddens, num_layers = len(vocab), 256, 2
num_inputs = vocab_size
lstm_layer = nn.LSTM(num_inputs, num_hiddens, num_layers, bidirectional=True) # 唯一的不同就是在这, bidirecitonal = True
model = d2l.RNNModel(lstm_layer, len(vocab))
model = model.to(device)
num_epochs, lr = 500, 1
d2l.train_ch8(model, train_iter, vocab, lr, num_epochs, device)
```
