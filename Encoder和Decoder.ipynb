```python
# 重新考察CNN
# CNN里放入一张图, 经过layers, 最后通过Softmax回归得到Output
# 编码器: 将输入编程成中间表达形式(特征), 也就是CNN里面中间layers的特征提取
# 解码器: 将中间表示解码成输出, 也就是把CNN的特征翻译成标号

# 重新考察RNN
# RNN里给入一个句子, 最后得到预测的输出
# 编码器: 输入句子的向量和中间的隐藏层
# 解码器: 最后在最后一个时刻通过fc, 得到的预测, 就是解码器

# 编码器-解码器架构
# 一个模型被分成两块: Encoder, Decoder
#  - Encoder: 拿到一个输入, 表示成一个中间的状态(feature, hidden state), 放入Decoder
#  - Decoder: 也可以拿一些额外的Input, 拿到一个Encoder的输入, 做输出
```

```python
from torch import nn

# 编码器
class Encoder(nn.Module):
    def __init__(self, **kwargs):
        super(Encoder, self).__init__(**kwargs)
        
    def forward(self, X, *args):
        raise NotImplementedError
        
# 解码器
class Decoder(nn.Module):
    def __init__(self, **kwargs):
        super(Decoder, self).__init__(**kwargs)
    
    def init_state(self, enc_outputs, *args): # 中间的状态: 编码器的东西怎么传给他
        raise NotImplementedError
        
    def forward(self, X, state) # 额外的输入和state
        raise NotImplementedError
        
# 合并编码器和解码器
class EncoderDecoder(nn.Module):
    def __init__(self, encoder, decoder, **kwargs):
        super(EncoderDecoder, self).__init__(**kwargs)
        self.encoder = encoder
        self.decoder = decoder
        
    def forward(self, enc_X, dec_X, *args):
        enc_outputs = self.encoder(enc_X, *args)
        dec_state = self.decoder.init_state(enc_outputs, *args)
        return self.decoder(dec_X, dec_state)
```
